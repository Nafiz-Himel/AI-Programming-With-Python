Filtering Based on String

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah"],
    "City": ["New York", "Los Angeles", "Newark", "Boston", "New Delhi", "Chicago", "New Orleans", "San Francisco"],
    "Department": ["HR", "IT", "Finance", "Marketing", "Finance", "HR", "IT", "Marketing"],
    "Salary": [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000]
}

df = pd.DataFrame(data)
df
```

```python
df.loc[df["City"].str.contains('New')]
```

```python
df.loc[df["City"].str.contains('o')]
```

```python
df.loc[df["City"].str.contains('new', case=False)]
```

```python
# start with Los
df.loc[df["City"].str.contains(r"^Los")]
```

```python
# end with rk
df.loc[df["City"].str.contains(r"rk$")]
```

```python
# name start with an vowel
df.loc[df["Name"].str.contains(r"^[AEIOU]")]
```

```python
# string contains New or Los
df.loc[df["City"].str.contains(r"New|Los")]
```
