Class & Object

```python
class Phone:
    category = "Electronics" ## class attribute 

    # constructor
    def __init__(self,model,bettery,camera,bettery_percentage=100):
        self.model = model
        self.battery = bettery
        self.camera = camera
        self.bettery_percentage = bettery_percentage

    #methods
    def charge(self,hour):
        self.charge += hour

    def capture(self,photo):
        if(self.bettery_percentage) <= 0:
            print("No charge")
        else: self.bettery_percentage -= photo
        print(f"Photo captured in {self.model}")
```

```python
## creating object

apple = Phone()
print(type(apple))
print(apple.category)

blueberry = Phone()
blueberry.category = "Super Electronics"
motorola = Phone()
print(blueberry.category)
print(motorola.category)
```

```python
samsung = Phone("Iphone17", 3000, 40)
honor = Phone("Honor17", 3500, 50)

print(honor.camera,honor.bettery_percentage)

honor.capture(10)
print(honor.bettery_percentage)
```
