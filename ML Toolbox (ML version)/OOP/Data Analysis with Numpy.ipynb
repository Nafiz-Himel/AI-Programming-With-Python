## Ndarray

```python
import numpy as np

# 1d array
arr1 = np.array([1,2,3,4,5])
print(type(arr1))
print(arr1.ndim)

# 2d array
arr2 = np.array([[1,2,3], #row1
           [4,5,6] #row2
           ])
print(arr2.ndim)


# 3d array
arr3 = np.array([
    #1st floor
    [
        [4,5,6], #row1
        [1,2,3]  #row2
    ],
    #2nd floor
    [
        [0,1,2], #row1
        [9,8,7]  #row2
    ]
])

print(arr3.ndim)
```

## Array  Attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

```python
#dictionary
dc = {'a' : 10,'b': 20,'c' : 3}
keys = dc.keys()
values = dc.values()
items = dc.items()


#keys to ndarray
arr = np.array(list(keys))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)

#values to ndarray
arr = np.array(list(values))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)

#items to ndarray
arr = np.array(list(items))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

indexing and slicing

```python
#indexing

print(arr1)
print(arr1[1])
arr1[1] = 100
print(arr1,arr1[1])

arr2 = np.array([ [1,2,3], # row1
                  [4,5,6] #row2
                ])
print(arr2)
print(arr2[0][1])
arr2[0][1] = 100
print(arr2[0][1])
print(arr2)
```

```python
# slicing
# 1d = arr[start:end:step]

arr1_mod = arr1[0:2]
print(arr1)
print(arr1_mod)

```

```python
#2d slicing
# arr [row_start :row_end: step, column_start: column end:step]
print(arr2)

# getting a row
row_0 = arr2[0:1,:]
row_1 = arr2[1:,:]
# print(row_0)
print(row_1)


#getting a column
column_0 = arr2[:,0:1]
column_1 = arr2[:,1:2]
print(column_0)
print(column_1)
```

```python
#getting a portion

portion = arr2[::, 0:2]
print(portion)
```

```python
#slicing
#1d = arr[start:end:step]

arr1_mod = arr1[0:2].copy()
arr1[1] = 100


arr1_mod[1] = 200
print(arr1_mod)
print(arr1)
```

ndarray Arithmetic Operators and Mathematical function

```python
# faster for vectorization
x = np.array([10 ,8 ,100, 40,])
y = np.array([2, 3, 4, 5])

add = x+y
# add = np.add(x,y)
print(add)

sub = x-y
print(sub)

mul = x*y
print(mul)

div = x/y
print(div)

rem = x%y
print(rem)
```

```python
# mathematical function

# trigonometry

# sin, cos, tan takes values in radian
sin_val = np.sin(x)
print(sin_val)

deg_cnvt = np.rad2deg(x)
print(deg_cnvt)

rad_cnvt = np.deg2rad(deg_cnvt)
print(rad_cnvt)
```

```python
# log

base_10 = np.log10(x)
print(base_10)

two_diye_bag = np.log2(x)
print(two_diye_bag)

sqrt = np.sqrt(x)
print(sqrt)
```

```python
# different

s = np.sum(x)
print(s)

#prefix sum er moto
cumul = np.cumulative_sum(x)
print(cumul)
```

Logical Funciton

```python
# comparison - >,< , >= ,<= , == , !=

x = np.array([ 10 , 8 , 16 , 100 ])
y = np.array([ 2 , 3 ,16  ,5 ])

greater_than = x>y

print(greater_than)

equal = x==y

print(equal)

# all () -> shob true kina
# any() -> jekono ekta true ase kina

print(np.all(greater_than))

print(np.any(equal))
```

Sorting

```python
#inplace
x = np.array([ 10 , 8 , 16 , 100 ])
z = x.copy()

print(z)
z.sort()
print(z)
print(x)
```
