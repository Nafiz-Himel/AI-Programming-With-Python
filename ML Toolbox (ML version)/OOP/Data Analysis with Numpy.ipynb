## Ndarray

```python
import numpy as np

# 1d array
arr1 = np.array([1,2,3,4,5])
print(type(arr1))
print(arr1.ndim)

# 2d array
arr2 = np.array([[1,2,3], #row1
           [4,5,6] #row2
           ])
print(arr2.ndim)


# 3d array
arr3 = np.array([
    #1st floor
    [
        [4,5,6], #row1
        [1,2,3]  #row2
    ],
    #2nd floor
    [
        [0,1,2], #row1
        [9,8,7]  #row2
    ]
])

print(arr3.ndim)
```

## Array  Attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

## ndarray data type

```python
arr = np.array([1,2,3])
print(arr.dtype)

arr = np.array([1,2,3,3.5,0.5])
print(arr.dtype)
```

```python
arr = np.array([1,2,3,3.5]) ## upcasted to float
print(arr.dtype)

arr = np.array([1,2,3,3.5,'3']) ## upcasted to string
print(arr.dtype)
```

### selecting a data type for an array

```python
arr = np.array([1,2,3],dtype=np.int8)
print(arr.dtype)

arr = np.array([1,2,300],dtype=np.int8) ## -127 -> +127
print(arr.dtype)
```

```python
arr = np.array([1,2,3])
print(arr.dtype)

arr = arr.astype(np.int16)
print(arr.dtype)

# arr = np.array([1,2,3,'hello'],dtype=np.float64)
```
