Different type of argument

```python
# positional arguments
def adderr(a,b):
    print(f"{a},{b}")


adderr(10,20)
```

```python
#Arbitrary Positional Arguments
def adder(*args):
    # print(args) ->tuple a dibe
    for i in args:
        print(i,type(i)) # int dibe
adder(1,2,5,3)
```

```python
# keyword argumnets
def ad(a,b,c):
    print(f"{a},{b},{c}")
ad(a=10,c=30,b=20)
```

```python
def addd(**kwargs):
    # print(kwargs , type(kwargs))
    for key,value in kwargs.items():
        print(f"{key}: {value}")
addd(a = 10, b = 30,n_estimator = 50)
```

Generator

```python
lst = [x for x in range(500)] #dataset

print(lst)

def data_loader(chunk_size,list):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]

x = data_loader(5,lst)

print(type(x))
print(x)
print(next(x))
print(next(x))
```
