2D line plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
hours = [2 , 3 , 4, 1 , 5 , 7 ,3] 
days = [1 ,2 , 3 ,4 ,5 ,6 ,7] 

plt.plot(days,hours)
```

```python
plt.xlabel("Days")
plt.ylabel("Study Hour")

hours = [2 , 3 , 4, 1 , 5 , 7 ,3] 
days = [1 ,2 , 3 ,4 ,5 ,6 ,7] 

plt.plot(days,hours)
```

```python
plt.xlabel("Days")
plt.ylabel("Study Hour")
plt.title("Trend of a student study time over a week")

hours = [2 , 3 , 4, 1 , 5 , 7 ,3] 
days = [1 ,2 , 3 ,4 ,5 ,6 ,7] 

plt.plot(days,hours)
```

```python
plt.xlabel("Days")
plt.ylabel("Study Hour")
plt.title("Trend of a student study time over a week")

hours_1 = [2 , 3 , 4, 1 , 5 , 7 ,3] 
hours_2 = [1 , 4 , 3, 5 , 7 , 7 ,1] 
days = [1 ,2 , 3 ,4 ,5 ,6 ,7] 

plt.plot(days,hours_1)
plt.plot(days,hours_2)
```
