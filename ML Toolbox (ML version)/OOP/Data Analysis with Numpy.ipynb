## Ndarray

```python
import numpy as np

# 1d array
arr1 = np.array([1,2,3,4,5])
print(type(arr1))
print(arr1.ndim)

# 2d array
arr2 = np.array([[1,2,3], #row1
           [4,5,6] #row2
           ])
print(arr2.ndim)


# 3d array
arr3 = np.array([
    #1st floor
    [
        [4,5,6], #row1
        [1,2,3]  #row2
    ],
    #2nd floor
    [
        [0,1,2], #row1
        [9,8,7]  #row2
    ]
])

print(arr3.ndim)
```

## Array  Attributes

```python
# dimension
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)

# shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

#data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

#size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

### ndarray creation from existing data

```python
#list

lst = [10,20,30,40,40.5]
# arr = np.array(lst)
arr = np.array(lst, dtype = np.int32)

print(arr)
print(type(arr))
print(arr.dtype)


mxd_lst = [10,True, 'Hello', 10.5]
arr = np.array(mxd_lst)

print(arr)
print(type(arr))
print(arr.dtype)


matrix = [
    [1,2,3],
    [4,5,6]
]

arr = np.array(matrix)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#tuple

tpl = (10,20,30)
arr = np.array(tpl)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#dictionary
dc = {'a' : 10,'b': 20,'c' : 3}
keys = dc.keys()
values = dc.values()
items = dc.items()


#keys to ndarray
arr = np.array(list(keys))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)

#values to ndarray
arr = np.array(list(values))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)

#items to ndarray
arr = np.array(list(items))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

array creation with random values

```python
# np.random.rand(shape)

arr = np.random.rand(2,3) # 0 ->1
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.randint(starting,ending,shape)

arr = np.random.randint(1,10,(2,3))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.uniform(starting,ending,shape)

arr = np.random.uniform(50,100,(2,3))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

ndarray creation from range

```python
#np.arange(first,end,step) -> similar to range function

arr = np.arange(1,10,1).reshape(3,3)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.linspace(start,end,koita)

arr = np.linspace(0,4,6)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#np.logpace(start,end(included),koita)

# arr = np.logspace(0,4,6)
arr = np.logspace(0,4,6,base=2)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

indexing and slicing

```python
#indexing

print(arr1)
print(arr1[1])
arr1[1] = 100
print(arr1,arr1[1])

arr2 = np.array([ [1,2,3], # row1
                  [4,5,6] #row2
                ])
print(arr2)
print(arr2[0][1])
arr2[0][1] = 100
print(arr2[0][1])
print(arr2)
```

```python
# slicing
# 1d = arr[start:end:step]

arr1_mod = arr1[0:2]
print(arr1)
print(arr1_mod)

```

```python
#2d slicing
# arr [row_start :row_end: step, column_start: column end:step]
print(arr2)

# getting a row
row_0 = arr2[0:1,:]
row_1 = arr2[1:,:]
# print(row_0)
print(row_1)


#getting a column
column_0 = arr2[:,0:1]
column_1 = arr2[:,1:2]
print(column_0)
print(column_1)
```

```python
#getting a portion

portion = arr2[::, 0:2]
print(portion)
```

```python
#slicing
#1d = arr[start:end:step]

arr1_mod = arr1[0:2].copy()
arr1[1] = 100


arr1_mod[1] = 200
print(arr1_mod)
print(arr1)
```
